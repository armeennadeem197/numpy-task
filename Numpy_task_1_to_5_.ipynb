# Student Marks Analyzer

Now we will use different NumPy concepts in one small program.

## Problem

We have marks of different students.

We want to find:

- Average marks
- Highest marks
- Lowest marks
- Total passed students
- Total failed students

`````python
import numpy as np

marks = np.array([45, 78, 90, 32, 67, 88, 55, 40, 95, 28])

print("Student Marks:")
print(marks)

print("\nAverage Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))

passed = marks[marks >= 50]
failed = marks[marks < 50]

print("\nPassed Students Marks:")
print(passed)

print("Failed Students Marks:")
print(failed)

print("\nTotal Passed:", len(passed))
print("Total Failed:", len(failed))
`````
